### __Web Scraping__

#### __Required Tools__

Two modules are required:
Requests and Beautiful Soup

```python
from bs4 import BeautifulSoup
import requests
```

#### __Fetching and parsing HTML__

To start web scraping, HTML content of a web page has to be fetched and parsed using Beautiful Soup.

```python
#Specify the URL of the webpage to be scraped
url = 'https://en.wikipedia.org/wiki/IBM'

#Send an HTTP GET request to the webpage
response = requests.get(url)

#Store the HTML content in a variable
html_content = response.text

#Create a BeautifulSoup object to parse the html
soup = BeautifulSoup(html_content, 'html.parser')

#Display a snippet of the HTML content
print(html_content[:500])
```

### __Navigating the HTML Structure__

BeautifulSoup represents HTML content as tree-like structure.
Methods:
* find_all - filter and extract specific HTML elements


```python
# Find all anchor tags <a> in the  HTML
links = soup.find_all('a')

#Iterate through the list of links and print their text
for link in links:
    print(link.text)
```
