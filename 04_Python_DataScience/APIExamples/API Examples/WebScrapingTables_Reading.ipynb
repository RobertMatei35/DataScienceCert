### Web Scraping Tables using Pandas

The Pandas library in Python contains a function read_html() that can be used to extract tabular information from any web page.

__Example__

We want to extract the list of the largest banks in the world by market capitalization using Wikipedia:
URL = "https://en.wikipedia.org/wiki/List_of_largest_banks"

__To extract all the tables in the web page, pandas.read_html() method can be used__


A snapshot of the page is shown below. The required table is the first one in the webpage
![Alt Text](../LargestBanksImage.png)


The following lines of code are required to extract the table from the webpage:


```python
import io
import pandas as pd
import requests

URL = "https://en.wikipedia.org/wiki/List_of_largest_banks"
header = {"User-Agent": "Mozilla/5.0"}


html = requests.get(URL, headers=header).text
tables = pd.read_html(io.StringIO(html))

print(type(tables))

df = tables[0]

print(df)
```
